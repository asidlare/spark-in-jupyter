# JDBC
https://spark.apache.org/docs/latest/sql-data-sources-jdbc.html

```python
PATH = "/code/notebooks/postgresql-42.2.18.jar"
HOST = "db"
USER = "postgres"
PASS = "postgres"
TABLE = "abcd"
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as f
```

```python
spark = SparkSession.builder.appName("postgres").config("spark.jars", PATH).getOrCreate()
```

```python
pg_views = spark.read \
                .format("jdbc") \
                .option("url", f"jdbc:postgresql://{HOST}:5432/postgres") \
                .option("dbtable", "pg_views") \
                .option("user", USER) \
                .option("password", PASS) \
                .option("driver", "org.postgresql.Driver") \
                .load()
```

```python
pg_views.show()
```

```python
df = spark.range(100).select("id", f.lit("a").alias("val"))
```

```python
df.show()
```

```python
df.write \
  .format("jdbc") \
  .option("url", f"jdbc:postgresql://{HOST}:5432/postgres") \
  .option("dbtable", TABLE) \
  .option("user", USER) \
  .option("password", PASS) \
  .option("driver", "org.postgresql.Driver") \
  .save()
```

```python
df_back = spark.read \
               .format("jdbc") \
               .option("url", f"jdbc:postgresql://{HOST}:5432/postgres") \
               .option("dbtable", TABLE) \
               .option("user", USER) \
               .option("password", PASS) \
               .option("driver", "org.postgresql.Driver") \
               .load()
```

```python
df_back.show()
```
